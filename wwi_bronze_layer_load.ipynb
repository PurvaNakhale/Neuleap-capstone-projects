```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark = SparkSession.builder.appName("World Wide Importers").getOrCreate()
```

```python
dbutils.widgets.text('catalog', '', 'Catalog')
dbutils.widgets.text('schema', '', 'Schema')
```

```python
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")

path = "/Volumes/raw_dataset/raw/raw_files/"
raw_files = dbutils.fs.ls(path)

for file in raw_files:
    df = spark.read.csv(file.path, header = True)
    table_name = file.name.replace('csv', '').replace('.', '').lower()
    # print(table_name)
    if df.isEmpty():
        raise ValueError(f"No data found in csv {file.name}")
    else:
        df = df.select([col(c).cast("string") for c in df.columns])
        df = df.withColumn('business_month', lit('June 2016')).withColumn('ingestion_timestamp', current_timestamp())
        df.write.saveAsTable(f'{catalog}.{schema}.{table_name}')
        print(f"Created table {table_name}")
```

```python

# Get all tables in the schema
tables = spark.sql(f"SHOW TABLES IN {catalog}.{schema}").collect()

# Drop each table
for table in tables:
    print(f"No. of rows in {table.tableName} : {spark.sql(f'select count(*) from {catalog}.{schema}.{table.tableName}').collect()[0][0]}")
```
